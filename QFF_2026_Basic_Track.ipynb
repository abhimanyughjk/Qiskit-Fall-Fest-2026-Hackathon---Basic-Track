```python
import pandas as pd
import numpy as np
from qiskit import QuantumCircuit, transpile

# Confirm that the organizer-provided models are loaded.
assert set(models.keys()) == {"2016", "contemporary", "hyperbolic"}

def make_bell_measurement(basis):
    """Prepare |Phi+> and measure XX, YY, or ZZ."""
    qc = QuantumCircuit(2, 2)

    qc.h(0)
    qc.cx(0, 1)

    if basis == "XX":
        qc.h(0)
        qc.h(1)
    elif basis == "YY":
        qc.sdg(0)
        qc.sdg(1)
        qc.h(0)
        qc.h(1)
    elif basis != "ZZ":
        raise ValueError("Unknown measurement basis")

    qc.measure([0, 1], [0, 1])
    return qc


def get_correlator(counts):
    """Calculate the parity correlation from two-bit outcomes."""
    total = sum(counts.values())
    even = sum(
        count for bits, count in counts.items()
        if sum(int(b) for b in bits.replace(" ", "")) % 2 == 0
    )
    odd = total - even
    return (even - odd) / total, total


def circuit_cost(circuit):
    """Count compiled two-qubit gates and explicit SWAP operations."""
    two_qubit = sum(
        len(item.qubits) == 2
        for item in circuit.data
        if item.operation.name not in ("barrier",)
    )
    swaps = sum(
        item.operation.name == "swap"
        for item in circuit.data
    )
    return circuit.depth(), two_qubit, swaps


def run_baseline(model, backend, label):
    correlators = {}
    compiled_circuits = []

    for basis in ("XX", "YY", "ZZ"):
        circuit = make_bell_measurement(basis)

        compiled = transpile(
            circuit,
            backend=backend,
            initial_layout=list(model.ports),
            optimization_level=1,
            seed_transpiler=42
        )
        compiled_circuits.append(compiled)

        result = backend.run(
            compiled,
            shots=model.shots,
            seed_simulator=42
        ).result()

        counts = result.get_counts()
        correlators[basis], _ = get_correlator(counts)

    xx = correlators["XX"]
    yy = correlators["YY"]
    zz = correlators["ZZ"]
    n = model.shots

    fidelity = (1 + xx - yy + zz) / 4

    sigma_xx = np.sqrt(max(0, 1 - xx**2) / n)
    sigma_yy = np.sqrt(max(0, 1 - yy**2) / n)
    sigma_zz = np.sqrt(max(0, 1 - zz**2) / n)

    sigma_f = np.sqrt(
        sigma_xx**2 + sigma_yy**2 + sigma_zz**2
    ) / 4

    # Report the maximum cost across the three measurement circuits.
    costs = [circuit_cost(c) for c in compiled_circuits]

    return {
        "Model": model.name,
        "Case": label,
        "Shots/setting": n,
        "XX": xx,
        "YY": yy,
        "ZZ": zz,
        "Fidelity": fidelity,
        "Sigma_F": sigma_f,
        "F - 2sigma": fidelity - 2 * sigma_f,
        "Verified": fidelity - 2 * sigma_f > 0.5,
        "Max depth": max(c[0] for c in costs),
        "Max 2-qubit gates": max(c[1] for c in costs),
        "Max explicit SWAPs": max(c[2] for c in costs),
    }


results = []

for name, model in models.items():
    print(f"\nRunning {name}...")

    results.append(
        run_baseline(model, model.ideal_backend, "Ideal")
    )
    results.append(
        run_baseline(model, model.backend, "Noisy")
    )

baseline_df = pd.DataFrame(results)

display(
    baseline_df.round(4).sort_values(["Model", "Case"])
)
```

```python
import rustworkx as rx

def make_protected_bell_measurement(basis):
    """Prepare |Phi+> with an ancilla for a ZZ parity check."""
    qc = QuantumCircuit(3, 3)

    # Prepare Bell state
    qc.h(0)
    qc.cx(0, 1)

    # ZZ Parity Check (CNOT from ports to ancilla)
    qc.cx(0, 2)
    qc.cx(1, 2)
    qc.measure(2, 2)

    # Measurement basis rotation for XX or YY
    if basis == "XX":
        qc.h(0)
        qc.h(1)
    elif basis == "YY":
        qc.sdg(0)
        qc.sdg(1)
        qc.h(0)
        qc.h(1)

    # Measure ports
    qc.measure([0, 1], [0, 1])
    return qc

def get_conditional_correlator(counts):
    """Filter out shots that fail the parity check (ancilla == 1)."""
    total_accepted = 0
    even = 0
    odd = 0

    for bitstring, count in counts.items():
        bitstring = bitstring.replace(" ", "")
        ancilla_bit = bitstring[0]
        port_bits = bitstring[1:]

        # Post-select: keep only shots where ancilla measured 0 (ZZ = +1)
        if ancilla_bit == '0':
            total_accepted += count
            parity = sum(int(b) for b in port_bits) % 2
            if parity == 0:
                even += count
            else:
                odd += count

    if total_accepted == 0:
        return 0, 0

    return (even - odd) / total_accepted, total_accepted

def run_protected(model, backend, label):
    correlators = {}
    compiled_circuits = []
    accepted_shots_list = []

    # Geometry-aware ancilla selection:
    # Pick the physical qubit that minimizes the sum of distances to port A and port B.
    dist_matrix = rx.distance_matrix(model.graph())
    port_a, port_b = model.ports

    best_ancilla = None
    min_dist = float('inf')

    for q in range(model.num_qubits):
        if q in (port_a, port_b):
            continue
        # Total distance from this qubit to both ports
        d = dist_matrix[q, port_a] + dist_matrix[q, port_b]
        if d < min_dist:
            min_dist = d
            best_ancilla = q

    for basis in ("XX", "YY", "ZZ"):
        circuit = make_protected_bell_measurement(basis)

        # Map logical qubits 0, 1, 2 to the ports and the optimal ancilla
        layout = [port_a, port_b, best_ancilla]

        compiled = transpile(
            circuit,
            backend=backend,
            initial_layout=layout,
            optimization_level=1,
            seed_transpiler=42
        )
        compiled_circuits.append(compiled)

        result = backend.run(
            compiled,
            shots=model.shots,
            seed_simulator=42
        ).result()

        counts = result.get_counts()
        corr, accepted = get_conditional_correlator(counts)
        correlators[basis] = corr
        accepted_shots_list.append(accepted)

    xx = correlators["XX"]
    yy = correlators["YY"]
    zz = correlators["ZZ"]

    # Use the minimum accepted shots across settings for a conservative uncertainty estimate
    n = min(accepted_shots_list)
    acceptance_fraction = n / model.shots

    fidelity = (1 + xx - yy + zz) / 4

    sigma_xx = np.sqrt(max(0, 1 - xx**2) / n) if n > 0 else 0
    sigma_yy = np.sqrt(max(0, 1 - yy**2) / n) if n > 0 else 0
    sigma_zz = np.sqrt(max(0, 1 - zz**2) / n) if n > 0 else 0
    sigma_f = np.sqrt(sigma_xx**2 + sigma_yy**2 + sigma_zz**2) / 4 if n > 0 else 0

    costs = [circuit_cost(c) for c in compiled_circuits]

    return {
        "Model": model.name,
        "Case": label,
        "Shots/setting": n,
        "XX": xx,
        "YY": yy,
        "ZZ": zz,
        "Fidelity": fidelity,
        "Sigma_F": sigma_f,
        "F - 2sigma": fidelity - 2 * sigma_f,
        "Verified": fidelity - 2 * sigma_f > 0.5 if n > 0 else False,
        "Max depth": max(c[0] for c in costs),
        "Max 2-qubit gates": max(c[1] for c in costs),
        "Max explicit SWAPs": max(c[2] for c in costs),
        "Yield (%)": round(acceptance_fraction * 100, 1)
    }

# Append the 3 protected cases to your existing results list
protected_results = []
for name, model in models.items():
    print(f"Running protected case for {name}...")
    protected_results.append(run_protected(model, model.backend, "Protected"))

# Display the final 9-case comparison
final_df = pd.DataFrame(results + protected_results)
final_df.fillna({"Yield (%)": 100.0}, inplace=True)
display(final_df.round(4).sort_values(["Model", "Case"]))
```
